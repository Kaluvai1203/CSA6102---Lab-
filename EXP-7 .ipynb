```python
# Dictionary of common file signatures (magic numbers)
FILE_SIGNATURES = {
    b'\xFF\xD8\xFF': 'JPEG Image (.jpg)',
    b'\x89PNG\r\n\x1A\n': 'PNG Image (.png)',
    b'GIF87a': 'GIF Image (.gif)',
    b'GIF89a': 'GIF Image (.gif)',
    b'%PDF': 'PDF Document (.pdf)',
    b'PK\x03\x04': 'ZIP Archive / DOCX / XLSX / PPTX (.zip)',
    b'ID3': 'MP3 Audio (.mp3)',
    b'RIFF': 'WAV/AVI File',
    b'\x7FELF': 'Linux Executable (ELF)',
    b'MZ': 'Windows Executable (.exe)'
}

def identify_file_type(file_path):
    try:
        with open(file_path, "rb") as file:
            header = file.read(16)  # Read first 16 bytes

        for signature, file_type in FILE_SIGNATURES.items():
            if header.startswith(signature):
                return file_type

        return "Unknown file type"

    except FileNotFoundError:
        return "File not found."


# Main Program
file_path = input("Enter the file path: ")

result = identify_file_type(file_path)

print("\nDetected File Type:", result)
```
