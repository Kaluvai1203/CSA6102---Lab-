```python
import hashlib
import shutil

# Function to calculate SHA-256 hash
def calculate_hash(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()


# Function to create a forensic copy
def create_forensic_copy(source_file, destination_file):
    shutil.copyfile(source_file, destination_file)
    print("Forensic copy created successfully.")


# Main Program
source = input("Enter the source file path: ")
destination = input("Enter the destination file path: ")

# Create forensic copy
create_forensic_copy(source, destination)

# Calculate hashes
original_hash = calculate_hash(source)
copied_hash = calculate_hash(destination)

# Display hashes
print("\nOriginal File SHA-256 Hash :", original_hash)
print("Copied File SHA-256 Hash   :", copied_hash)

# Verify integrity
if original_hash == copied_hash:
    print("\nIntegrity Verification: SUCCESS")
    print("The forensic copy is an exact bit-stream copy.")
else:
    print("\nIntegrity Verification: FAILED")
    print("The copied file has been altered or corrupted.")
```
