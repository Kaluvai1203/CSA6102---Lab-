```python
from collections import defaultdict

THRESHOLD = 5

failed_attempts = defaultdict(int)

log_file = input("Enter login log file name: ")

try:
    with open(log_file, "r") as file:

        for line in file:
            parts = line.split()

            status = parts[2]
            ip = parts[3]

            if status == "FAILED":
                failed_attempts[ip] += 1

    print("\nBrute Force Detection Report")
    print("-" * 35)

    found = False

    for ip, count in failed_attempts.items():
        if count >= THRESHOLD:
            print(f"Suspicious IP: {ip}")
            print(f"Failed Login Attempts: {count}")
            print("Status: Possible Brute-Force Attack\n")
            found = True

    if not found:
        print("No brute-force attacks detected.")

except FileNotFoundError:
    print("Log file not found.")
```
