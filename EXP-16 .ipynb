```python
import csv

ransomware_data = [
    {
        "Family": "WannaCry",
        "Domain": "iuqerfsodp9ifjaposdfjhgosurijfaewrwergwea.com",
        "IP": "192.168.1.10",
        "Extension": ".WNCRY",
        "Note": "Oops, your files have been encrypted!"
    },
    {
        "Family": "Locky",
        "Domain": "locky-payment.com",
        "IP": "203.0.113.10",
        "Extension": ".locky",
        "Note": "Your files are encrypted. Pay Bitcoin."
    },
    {
        "Family": "Ryuk",
        "Domain": "ryuk-help.net",
        "IP": "198.51.100.25",
        "Extension": ".RYK",
        "Note": "All your files have been encrypted."
    }
]

with open("ransomware_database.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow([
        "Family",
        "Domain",
        "IP",
        "Extension",
        "Ransom Note"
    ])

    for item in ransomware_data:
        writer.writerow([
            item["Family"],
            item["Domain"],
            item["IP"],
            item["Extension"],
            item["Note"]
        ])

print("Ransomware family database created successfully.\n")

print("========== IoC Indicators ==========\n")

for item in ransomware_data:

    print("Family      :", item["Family"])
    print("Domain      :", item["Domain"])
    print("IP Address  :", item["IP"])
    print("Extension   :", item["Extension"])
    print("Ransom Note :", item["Note"])
    print()

print("========== YARA-like Detection Rules ==========\n")

for item in ransomware_data:

    family = item["Family"]

    print(f"rule {family}_Detection")
    print("{")
    print("    strings:")
    print(f'        $domain = "{item["Domain"]}"')
    print(f'        $ext = "{item["Extension"]}"')
    print(f'        $note = "{item["Note"]}"')
    print()
    print("    condition:")
    print("        any of them")
    print("}")
    print()

print("========== Simulated Deployment Timeline ==========\n")

timeline = [
    "09:00  Initial phishing email received",
    "09:05  Malicious attachment opened",
    "09:08  Malware executed",
    "09:12  Files encrypted",
    "09:15  Ransom note created",
    "09:18  Communication with attacker infrastructure detected"
]

for event in timeline:
    print(event)

print("\n========== Summary ==========\n")

print("Total Families :", len(ransomware_data))

extensions = [x["Extension"] for x in ransomware_data]
domains = [x["Domain"] for x in ransomware_data]
ips = [x["IP"] for x in ransomware_data]

print("Extensions :", len(extensions))
print("Domains    :", len(domains))
print("IP Addresses :", len(ips))

print("\nDatabase saved as ransomware_database.csv")
```
