```python
case_number = input("Enter Case Number: ")
investigator = input("Enter Investigator Name: ")
date = input("Enter Investigation Date: ")
device = input("Enter Device Examined: ")
evidence = input("Enter Evidence Collected: ")
analysis = input("Enter Analysis Findings: ")
result = input("Enter Investigation Result: ")
recommendation = input("Enter Recommendations: ")

report = f"""
====================================================
        DIGITAL FORENSIC INVESTIGATION REPORT
====================================================

1. Case Information
-------------------
Case Number        : {case_number}
Investigator       : {investigator}
Investigation Date : {date}

2. Evidence Details
-------------------
Device Examined    : {device}
Evidence Collected : {evidence}

3. Investigation Findings
--------------------------
{analysis}

4. Result
----------
{result}

5. Recommendations
------------------
{recommendation}

====================================================
End of Report
====================================================
"""

with open("Forensic_Report.txt", "w") as file:
    file.write(report)

print("\nReport Generated Successfully!\n")
print(report)
print("\nThe report has been saved as 'Forensic_Report.txt'.")
```
