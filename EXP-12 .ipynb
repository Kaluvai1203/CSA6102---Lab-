```python
# Program: Extract and Display EXIF Metadata from an Image

from PIL import Image
from PIL.ExifTags import TAGS

# Step 1: Create a sample image
image = Image.new("RGB", (200, 200), color="blue")
image.save("sample.jpg")

print("Sample image 'sample.jpg' created successfully.\n")

# Step 2: Read image name
image_file = input("Enter image file name: ")

try:
    img = Image.open(image_file)

    exif_data = img.getexif()

    print("\n========== EXIF Metadata ==========")

    if exif_data:
        for tag_id, value in exif_data.items():
            tag = TAGS.get(tag_id, tag_id)
            print(f"{tag}: {value}")
    else:
        print("No EXIF metadata found in this image.")

except FileNotFoundError:
    print("Image file not found.")

except Exception as e:
    print("Error:", e)
```
