```python
with open("storage.bin", "wb") as f:
    f.write(b"RandomData12345")
    f.write(b"\xFF\xD8\xFF")          # JPEG signature
    f.write(b"SomeDeletedImageData")
    f.write(b"%PDF")                  # PDF signature
    f.write(b"DeletedPDFContent")
    f.write(b"\x89PNG\r\n\x1A\n")     # PNG signature
    f.write(b"DeletedPNGContent")
```

```python
# Simulate deleted file recovery by scanning for file signatures

FILE_SIGNATURES = {
    b'\xFF\xD8\xFF': 'JPEG Image',
    b'\x89PNG\r\n\x1A\n': 'PNG Image',
    b'%PDF': 'PDF Document',
    b'PK\x03\x04': 'ZIP Archive'
}

def scan_storage(file_path):
    try:
        with open(file_path, "rb") as file:
            data = file.read()

        found = False

        print("\nScanning storage for deleted file signatures...\n")

        for signature, file_type in FILE_SIGNATURES.items():
            position = data.find(signature)

            while position != -1:
                print(f"Possible {file_type} found at byte offset: {position}")
                found = True
                position = data.find(signature, position + 1)

        if not found:
            print("No recognizable deleted file signatures found.")

    except FileNotFoundError:
        print("Storage image file not found.")


# Main Program
storage_file = input("Enter storage image file name: ")
scan_storage(storage_file)
```
