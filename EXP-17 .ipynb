```python
import re

print("==============================================")
print(" Ethical Hacking Attack Pattern Analyzer")
print("==============================================")

attack_logs = [
    {"Type":"Brute Force","Username":"admin","Password":"123456"},
    {"Type":"Brute Force","Username":"admin","Password":"password"},
    {"Type":"Brute Force","Username":"admin","Password":"admin"},
    {"Type":"Brute Force","Username":"admin","Password":"welcome"},
    {"Type":"Brute Force","Username":"admin","Password":"Admin@123"},
    {"Type":"SQL Injection","Payload":"' OR '1'='1"},
    {"Type":"SQL Injection","Payload":"admin'--"},
    {"Type":"Directory Traversal","Payload":"../../etc/passwd"},
    {"Type":"Directory Traversal","Payload":"../../../windows/system32"},
]
weak_passwords = [
    "123456",
    "password",
    "admin",
    "welcome",
    "12345",
    "qwerty"
]

print("\n========== Weak Password Analysis ==========")

weak_count = 0

for log in attack_logs:

    if log["Type"] == "Brute Force":

        pwd = log["Password"]

        if pwd.lower() in weak_passwords:

            print("Weak Password Found :", pwd)
            weak_count += 1

print("\n========== SQL Injection Detection ==========")

sql_patterns = [
    r"'.*or.*'.*='",
    r"--",
    r";",
    r"union",
    r"select"
]

sql_count = 0

for log in attack_logs:

    if log["Type"] == "SQL Injection":

        payload = log["Payload"]

        detected = False

        for pattern in sql_patterns:

            if re.search(pattern, payload, re.IGNORECASE):

                detected = True

        if detected:

            print("Suspicious Payload :", payload)
            sql_count += 1

print("\n========== Directory Traversal Detection ==========")

traversal_count = 0

for log in attack_logs:

    if log["Type"] == "Directory Traversal":

        payload = log["Payload"]

        if "../" in payload or "..\\" in payload:

            print("Traversal Attempt :", payload)
            traversal_count += 1

print("\n========== Attack Pattern Signatures ==========")

print("BRUTE_FORCE : Multiple login attempts using common passwords")
print("SQL_INJECTION : OR, UNION, SELECT, --, ;")
print("DIRECTORY_TRAVERSAL : ../ or ..\\")

print("\n========== Vulnerability Report ==========")

if weak_count > 0:
    print("Weak Password Vulnerability : HIGH")

if sql_count > 0:
    print("SQL Injection Risk : HIGH")

if traversal_count > 0:
    print("Directory Traversal Risk : HIGH")

print("\n========== Suggested WAF Rules ==========")

print("1. Block requests containing '../'")
print("2. Block SQL keywords like UNION, SELECT, OR")
print("3. Block '--' and ';' characters")
print("4. Enable login rate limiting")
print("5. Use strong password policies")
print("6. Enable Multi-Factor Authentication")

print("\n========== Summary ==========")

print("Weak Passwords Detected :", weak_count)
print("SQL Injection Attempts  :", sql_count)
print("Traversal Attempts      :", traversal_count)

print("\nSimulation Completed Successfully.")
```
