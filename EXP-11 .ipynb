```python
from collections import defaultdict

with open("network_log.txt", "w") as f:
    f.write("""192.168.1.100 192.168.1.1 21
192.168.1.100 192.168.1.1 22
192.168.1.100 192.168.1.1 23
192.168.1.100 192.168.1.1 25
192.168.1.100 192.168.1.1 53
192.168.1.100 192.168.1.1 80
192.168.1.100 192.168.1.1 110
192.168.1.100 192.168.1.1 143
192.168.1.100 192.168.1.1 443
192.168.1.100 192.168.1.1 8080
192.168.1.101 192.168.1.1 80
192.168.1.101 192.168.1.1 443
""")

print("Sample network traffic log created successfully.\n")

log_file = input("Enter network log file name: ")

THRESHOLD = 5

source_ports = defaultdict(set)

try:
    with open(log_file, "r") as file:

        for line in file:
            parts = line.split()

            source_ip = parts[0]
            destination_port = parts[2]

            source_ports[source_ip].add(destination_port)

    print("\n========== Port Scan Detection Report ==========")

    found = False

    for ip, ports in source_ports.items():
        if len(ports) >= THRESHOLD:
            print(f"\nSuspicious Source IP : {ip}")
            print("Ports Scanned        :", sorted(ports))
            print("Total Ports          :", len(ports))
            print("Status               : Possible Port Scanning")
            found = True

    if not found:
        print("No port scanning activity detected.")

except FileNotFoundError:
    print("Network log file not found.")
```
