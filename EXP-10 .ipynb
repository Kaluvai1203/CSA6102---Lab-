```python
with open("email_header.txt", "w") as f:
    f.write("""From: support@paypal.com
Return-Path: hacker@gmail.com
Received: from unknown (192.168.10.50)
Received: from mail.fake.com (203.0.113.25)
Subject: Account Verification Required
""")

print("Sample email header file created successfully.\n")

file_name = input("Enter email header file name: ")

try:
    with open(file_name, "r") as file:
        headers = file.readlines()

    from_email = ""
    return_path = ""
    received = []

    for line in headers:
        if line.startswith("From:"):
            from_email = line.split(":", 1)[1].strip()

        elif line.startswith("Return-Path:"):
            return_path = line.split(":", 1)[1].strip()

        elif line.startswith("Received:"):
            received.append(line.strip())

    print("\n========== Email Header Analysis ==========")
    print("From Address :", from_email)
    print("Return-Path  :", return_path)

    print("\nReceived Headers:")
    for r in received:
        print(r)

    print("\n========== Spoofing Analysis ==========")

    spoof = False

    if from_email and return_path:
        from_domain = from_email.split("@")[-1]
        return_domain = return_path.split("@")[-1]

        if from_domain != return_domain:
            print("Warning: Sender domain and Return-Path domain do not match.")
            spoof = True

    for r in received:
        if "unknown" in r.lower():
            print("Warning: Suspicious 'Received' header detected.")
            spoof = True

    if spoof:
        print("\nResult: Possible Email Spoofing Detected.")
    else:
        print("\nResult: No spoofing indicators found.")

except FileNotFoundError:
    print("Header file not found.")
```
