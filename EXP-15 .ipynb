```python
import hashlib
import csv
import os


with open("clean_file.txt", "w") as f:
    f.write("This is a normal text file.")

with open("malware_sample.txt", "w") as f:
    f.write("This file simulates malware.")

print("Sample files created successfully.\n")

def calculate_hashes(filename):

    md5 = hashlib.md5()
    sha1 = hashlib.sha1()
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:

        while True:

            data = file.read(4096)

            if not data:
                break

            md5.update(data)
            sha1.update(data)
            sha256.update(data)

    return (
        md5.hexdigest(),
        sha1.hexdigest(),
        sha256.hexdigest()
    )


database = []

malware_md5, malware_sha1, malware_sha256 = calculate_hashes("malware_sample.txt")

database.append([
    "malware_sample.txt",
    malware_md5,
    malware_sha1,
    malware_sha256,
    "Trojan"
])

with open("malware_database.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerow([
        "File",
        "MD5",
        "SHA1",
        "SHA256",
        "Family"
    ])

    writer.writerows(database)

print("Malware database created.\n")

filename = input("Enter file name to scan: ")

md5, sha1, sha256 = calculate_hashes(filename)

print("\n========== HASH VALUES ==========")
print("MD5    :", md5)
print("SHA1   :", sha1)
print("SHA256 :", sha256)

detected = False

with open("malware_database.csv") as file:

    reader = csv.DictReader(file)

    for row in reader:

        if row["SHA256"] == sha256:

            print("\nMalware Detected!")
            print("Family :", row["Family"])
            detected = True

if not detected:
    print("\nNo malware detected.")

print("\n========== Detection Statistics ==========")

total = 2

malware = 1 if detected else 0

print("Files Scanned :", total)
print("Malware Found :", malware)

rate = malware / total * 100

print("Detection Rate: {:.2f}%".format(rate))

print("\n========== Hash Family Clustering ==========")

families = {}

with open("malware_database.csv") as file:

    reader = csv.DictReader(file)

    for row in reader:

        family = row["Family"]

        if family not in families:
            families[family] = 0

        families[family] += 1

for family, count in families.items():

    print(family, ":", count, "sample(s)")
```
